```python
# import statements
import pandas as pd
import re
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
# Initialize Tools
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

```python
def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"http\S+", "", text)          # remove URLs
    text = re.sub(r"<.*?>", "", text)            # remove HTML
    text = re.sub(r"\s+", " ", text)             # remove extra spaces
    return text.strip()

```

Tokenization

```python
def tokenize_text(text):
    return re.findall(r'\b\w+\b', text)

```

Stopword Removal

```python
def remove_stopwords(tokens):
    return [word for word in tokens if word not in stop_words]
```

Removing Non-Alphabetic Characters

```python
def remove_non_alpha(tokens):
    return [word for word in tokens if word.isalpha()]
```

Lemmatization

```python
def apply_lemmatization(tokens):
    return [lemmatizer.lemmatize(word) for word in tokens]
```

Handling Negations

```python
def handle_negations(tokens):
    negation_words = {"not", "no", "never", "n't"}
    new_tokens = []
    negate = False

    for word in tokens:
        if word in negation_words:
            negate = True
            continue

        if negate:
            new_tokens.append("not_" + word)
            negate = False
        else:
            new_tokens.append(word)

    return new_tokens
```

```python
def preprocess_for_ml(text):
    text = clean_text(text)
    tokens = tokenize_text(text)
    tokens = remove_non_alpha(tokens)
    tokens = handle_negations(tokens)  
    tokens = remove_stopwords(tokens)
    tokens = apply_lemmatization(tokens)
    return " ".join(tokens)

```

```python
def preprocess_for_dl(text):
    text = clean_text(text)
    return text
```

```python
def preprocess_for_transformer(text):
    text = clean_text(text)
    return text
```

```python
def generate_processed_columns(df, text_column="review_text"):
    # Generate ML processed text
    df["text_ml"] = df[text_column].apply(preprocess_for_ml)
    # Generate DL processed text
    df["text_dl"] = df[text_column].apply(preprocess_for_dl)
    # Generate Transformer processed text
    df["text_transformer"] = df[text_column].apply(preprocess_for_transformer)
    return df
```

```python
train_df = pd.read_csv("../data/processed/train.csv")
val_df = pd.read_csv("../data/processed/val.csv")
test_df = pd.read_csv("../data/processed/test.csv")
```

```python
train_df = generate_processed_columns(train_df, text_column="review_text")
val_df = generate_processed_columns(val_df, text_column="review_text")
test_df = generate_processed_columns(test_df, text_column="review_text")
```

```python
train_df.to_csv("../data/processed/train_all_versions.csv", index=False)
val_df.to_csv("../data/processed/val_all_versions.csv", index=False)
test_df.to_csv("../data/processed/test_all_versions.csv", index=False)

print("All splits processed and saved.")

```
